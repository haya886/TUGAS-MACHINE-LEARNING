*dataset ini diambil dari laman https://www.kaggle.com/datasets/mujtabamatin/air-quality-and-pollution-assessment. dataset ini digunakan untuk klasifikasi multiclass, di mana setiap rekam medis/stasiun pemantauan udara dikelompokkan ke dalam 1 kategori kualitas udara secara eksklusif.
